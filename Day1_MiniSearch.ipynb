```python
facts = [
    "SampleTech was founded in 2010.",
    "The company builds mobile apps for iOS and Android.",
    "The company offers AI and machine learning services.",
    "The head office is in Lahore.",
    "The team has more than 100 engineers."
]

stop_words = ["the", "is", "a", "an", "of", "in", "does", "do",
              "what", "where", "how", "for", "and", "to", "are", "it"]

question = input("Ask a question: ")
q_words = question.lower().replace("?", "").split()

best_fact = ""
best_score = 0

for fact in facts:
    fact_words = fact.lower().replace(".", "").split()
    score = 0
    for word in q_words:
        if word in stop_words:
            continue
        if word in fact_words:
            score += 1
    if score > best_score:
        best_score = score
        best_fact = fact

if best_score == 0:
    print("I don't have enough information.")
else:
    print("Answer:", best_fact)
```
